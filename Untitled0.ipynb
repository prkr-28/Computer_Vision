```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("images.jpg")
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Blur to remove noise
blurred = cv2.GaussianBlur(gray, (7, 7), 0)

# Noise map
noise_map = cv2.absdiff(gray, blurred)
noise_map_norm = cv2.normalize(noise_map, None, 0, 255, cv2.NORM_MINMAX)

# Display
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.subplot(1,3,2)
plt.title("Blurred")
plt.imshow(blurred, cmap='gray')
plt.axis('off')

plt.subplot(1,3,3)
plt.title("Noise Map")
plt.imshow(noise_map_norm, cmap='hot')
plt.axis('off')

plt.show()
```
